```python
query = """
SELECT
    City,
    ROUND(AVG(Price), 2) AS Average_Listing_Price
FROM listings
GROUP BY City
ORDER BY Average_Listing_Price DESC;
"""

cursor.execute(query)

rows = cursor.fetchall()

for row in rows:
    print(row)
```

```python
# What is the average price per square foot by property type?
```

```python
query = """
SELECT
    Property_Type,
    ROUND(AVG(Price / Sqft),2) AS Avg_Price_Per_Sqft
FROM listings
GROUP BY Property_Type
ORDER BY Avg_Price_Per_Sqft DESC;
"""

cursor.execute(query)

rows = cursor.fetchall()

for row in rows:
    print(row)
```

```python
query = """
SELECT
    p.Furnishing_Status,
    ROUND(AVG(l.Price),2) AS Avg_Price
FROM property p
JOIN listings l
ON p.Listing_ID = l.Listing_ID
GROUP BY p.Furnishing_Status;
"""

cursor.execute(query)

rows = cursor.fetchall()

for row in rows:
    print(row)
```

```python
# 4.Do properties closer to metro stations command higher prices?
```

```python
import pandas as pd

# Merge the DataFrames
df = pd.merge(
    df_listings,
    df3_property,
    left_on="Listing_ID",
    right_on="listing_id"
)

# Create Metro Distance Category
df["Metro_Category"] = df["metro_distance_km"].apply(
    lambda x: "Near Metro (<=2 km)" if x <= 2 else "Far from Metro (>2 km)"
)

# Calculate Average Price
result = (
    df.groupby("Metro_Category", as_index=False)["Price"]
      .mean()
      .round(2)
)

print(result)
```

```python
# 5.Are rented properties priced differently from non-rented ones?
```

```python
import pandas as pd

# Merge the DataFrames
df = pd.merge(
    df_listings,
    df3_property,
    left_on="Listing_ID",
    right_on="listing_id"
)

# Convert 0/1 into readable labels
df["Rental_Status"] = df["is_rented"].map({
    1: "Rented",
    0: "Not Rented"
})

# Calculate Average Price
result = (
    df.groupby("Rental_Status", as_index=False)["Price"]
      .mean()
      .round(2)
      .sort_values(by="Price", ascending=False)
)

print(result)
```

```python
# 10.How are properties distributed across price buckets?
```

```python
bins = [0, 500000, 1000000, 1500000, 2000000, float("inf")]

labels = [
    "Below 5 Lakhs",
    "5L - 10L",
    "10L - 15L",
    "15L - 20L",
    "Above 20L"
]

df["Price_Bucket"] = pd.cut(
    df["Price"],
    bins=bins,
    labels=labels
)

result = (
    df.groupby("Price_Bucket")
      .size()
      .reset_index(name="Property_Count")
)

print(result)
```
